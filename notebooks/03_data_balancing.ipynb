```python
import pandas as pd
from collections import Counter

df = pd.read_csv("../data/processed/grievances_cleaned.csv")
print("Label distribution:")
print(df['department'].value_counts())

# Check for any null values
print("\nNull values:", df.isnull().sum().sum())

# Verify labels are correct
valid_depts = ['Water Supply','Electricity','Roads','Sanitation','Transport']
invalid = df[~df['department'].isin(valid_depts)]
print("\nInvalid labels found:", len(invalid))
```

```python
df_clean = df.dropna(subset=['cleaned','department'])
df_clean.to_csv("../data/processed/train_final.csv", index=False)
print("Final training file saved:", df_clean.shape)
```
